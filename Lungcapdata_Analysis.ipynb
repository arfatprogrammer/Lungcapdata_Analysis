```python
# 1. IMPORT LIBRARIES
# -----------------------------
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# 2. LOAD DATASET
# -----------------------------
# Read Excel file
import pandas as pd

df = pd.read_csv("Lungcapdata.csv")
```

```python
# 3. DISPLAY DATA
# -----------------------------
print("\n================ FIRST 5 ROWS ================\n")
print(df.head())

print("\n================ LAST 5 ROWS ================\n")
print(df.tail())
```

```python
# 4. SHAPE OF DATASET
# -----------------------------
print("\n================ SHAPE ================\n")
print("Rows :", df.shape[0])
print("Columns :", df.shape[1])
```

```python
# 5. COLUMN NAMES
# -----------------------------
print("\n================ COLUMN NAMES ================\n")
print(df.columns)
```

```python
# 6. DATA TYPES
# -----------------------------
print("\n================ DATA TYPES ================\n")
print(df.dtypes)
```

```python
# 7. DATASET INFORMATION
# -----------------------------
print("\n================ DATASET INFO ================\n")
print(df.info())
```

```python
# 8. STATISTICAL SUMMARY
# -----------------------------
print("\n================ STATISTICAL SUMMARY ================\n")
print(df.describe())
```

```python
# 9. CHECK NULL VALUES
# -----------------------------
print("\n================ MISSING VALUES ================\n")
print(df.isnull().sum())
```

```python
# 10. CHECK DUPLICATES
# -----------------------------
print("\n================ DUPLICATE VALUES ================\n")
print("Total Duplicate Rows :", df.duplicated().sum())
```

```python
# 11. NUMERICAL & CATEGORICAL COLUMNS
# -----------------------------
num_cols = df.select_dtypes(include=np.number).columns
cat_cols = df.select_dtypes(exclude=np.number).columns

print("\nNumerical Columns :\n", num_cols)
print("\nCategorical Columns :\n", cat_cols)
```

```python
# VISUALIZATION SECTION
# =========================================================

# -----------------------------
# 12. HISTOGRAMS
# -----------------------------
df[num_cols].hist(figsize=(12,8))

plt.suptitle("Histograms of Numerical Features")

plt.tight_layout()

plt.show()
```

```python
# 13. BOXPLOTS
# -----------------------------
for col in num_cols:

    plt.figure(figsize=(6,4))

    plt.boxplot(df[col].dropna())

    plt.title(f"Boxplot of {col}")

    plt.ylabel(col)

    plt.show()
```

```python
# 14. LINE PLOTS
# -----------------------------
for col in num_cols:

    plt.figure(figsize=(8,4))

    plt.plot(df[col])

    plt.title(f"Line Plot of {col}")

    plt.xlabel("Index")

    plt.ylabel(col)

    plt.show()
```

```python
# 15. BAR CHARTS
# -----------------------------
for col in cat_cols:

    plt.figure(figsize=(7,5))

    df[col].value_counts().plot(kind='bar')

    plt.title(f"Bar Chart of {col}")

    plt.xlabel(col)

    plt.ylabel("Count")

    plt.show()
```

```python
# 16. PIE CHARTS
# -----------------------------
for col in cat_cols:

    plt.figure(figsize=(6,6))

    df[col].value_counts().plot(
        kind='pie',
        autopct='%1.1f%%'
    )

    plt.ylabel("")

    plt.title(f"Pie Chart of {col}")

    plt.show()
```

```python
# 17. SCATTER PLOT
# -----------------------------
plt.figure(figsize=(7,5))

plt.scatter(df['Age'], df['LungCap'])

plt.xlabel("Age")

plt.ylabel("Lung Capacity")

plt.title("Age vs Lung Capacity")

plt.show()

```

```python
# 18. HEIGHT VS LUNGCAP
# -----------------------------
plt.figure(figsize=(7,5))

plt.scatter(df['Height'], df['LungCap'])

plt.xlabel("Height")

plt.ylabel("Lung Capacity")

plt.title("Height vs Lung Capacity")

plt.show()
```

```python
# 19. CORRELATION MATRIX
# -----------------------------
corr = df[num_cols].corr()

plt.figure(figsize=(8,6))

plt.imshow(corr)

plt.colorbar()

plt.xticks(range(len(corr.columns)),
           corr.columns,
           rotation=90)

plt.yticks(range(len(corr.columns)),
           corr.columns)

plt.title("Correlation Matrix")

plt.show()

print("\n================ CORRELATION MATRIX ================\n")
print(corr)
```

```python
# 20. OUTLIER DETECTION
# -----------------------------
print("\n================ OUTLIERS ================\n")

for col in num_cols:

    Q1 = df[col].quantile(0.25)

    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)

    upper = Q3 + (1.5 * IQR)

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(f"{col} : {len(outliers)} outliers")
```

```python
# 21. MEAN, MEDIAN, MODE
# -----------------------------
print("\n================ CENTRAL TENDENCY ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Mean   :", df[col].mean())

    print("Median :", df[col].median())

    print("Mode   :", df[col].mode()[0])
```

```python
# 22. STANDARD DEVIATION
# -----------------------------
print("\n================ DISPERSION ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Standard Deviation :", df[col].std())

    print("Variance           :", df[col].var())
```

```python
# 23. SKEWNESS & KURTOSIS
# -----------------------------
print("\n================ DISTRIBUTION ANALYSIS ================\n")

for col in num_cols:

    print(f"\nColumn : {col}")

    print("Skewness :", df[col].skew())

    print("Kurtosis :", df[col].kurt())
```

```python
# 24. GROUP ANALYSIS
# -----------------------------
print("\n================ GROUP ANALYSIS ================\n")

print(df.groupby('Gender')['LungCap'].mean())

print("\n")

print(df.groupby('Smoke')['LungCap'].mean())
```
